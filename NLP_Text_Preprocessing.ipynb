# NLP Text Preprocessing Assignment

**Focus:** Clean raw text and build a preprocessing pipeline (no vectorization / ML models).

**Dataset:** 25 sample product reviews (synthetic).  
Style similar to Kaggle product-review datasets.  
Source: Created for this assignment (educational sample data).


## Setup


```python
import re
import pandas as pd

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

print('Setup done')
```

## Dataset


```python
raw_texts = [
    "This phone is AMAZING!!! Battery lasts 2 days. Buy at https://shop.example.com now!!!",
    "Worst laptop ever... screen broke in 3 weeks. Contact support@brand.com for help.",
    "<b>Great headphones</b> sound quality is superb. Price was $49 only.",
    "I love this camera. Photos are sharp. Visit www.photos.com for more reviews.",
    "The shoes are comfortable but size is small. Ordered 2 pairs yesterday.",
    "DO NOT BUY!!! Fake product. Email me at user123@gmail.com if you need proof.",
    "Nice watch delivery was fast. Tracking: https://track.ship.com/12345",
    "Book is okay... some pages missing. Rating: 3/5 stars.",
    "Amazing blender!!! Makes smoothies in 30 seconds. <p>Highly recommend</p>",
    "Tablet freezes often. Updated software 2 times still same issue",
    "The bag looks premium. Material quality is good for the price of 799 rs.",
    "Keyboard keys are sticky. Returned it on 15th March 2024.",
    "Fantastic service from seller. Got refund within 48 hours. Thanks!!!",
    "Perfume smell lasts only 1 hour. Not worth 2500 rupees.",
    "Mouse is wireless and smooth. Battery life is about 6 months.",
    "T-shirt fabric is soft but color faded after 2 washes.",
    "Check my full review at http://blog.reviews.net/item99 #gadget",
    "Chair is ergonomic. Back pain reduced in 1 week of use.",
    "Power bank capacity is less than advertised (10000mAh shown).",
    "Sunglasses look stylish. UV protection works well outdoors.",
    "The app crashes on Android 12. Contact appsupport@company.org",
    "Coffee maker is easy to clean. Makes 4 cups in 5 minutes.",
    "Speaker bass is strong but connectivity drops sometimes.",
    "Notebook quality is average. Pages are thin. Price: Rs.120",
    "This product is soooo good gr8 quality u should buy it now!!!"
]

df = pd.DataFrame({'id': range(1, 26), 'raw_text': raw_texts})
print('Total samples:', len(df))
print(df.head())
```

## PART 1 — NLP Pipeline & Basic Text Cleaning

### Task 1: Understanding Raw Text Data


```python
print('First 5 samples:')
print(df['raw_text'].head())

df['length'] = df['raw_text'].str.len()
print('\nLength of each text:')
print(df[['id', 'length']])

print('\nCommon issues in raw text:')
print('- Uppercase / lowercase mix')
print('- Punctuation (!!!, ...)')
print('- Numbers and prices')
print('- Extra spaces')
print('- URLs, emails, HTML tags (seen in samples)')
```

### Task 2: Basic Text Cleaning


```python
def basic_clean(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)   # remove punctuation and numbers
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['clean_text_basic'] = df['raw_text'].apply(basic_clean)

# side by side
print(df[['raw_text', 'clean_text_basic']].head(5))
```

## PART 2 — Advanced Text Cleaning

### Task 3: Removing Noise


```python
def advanced_clean(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\.\S+', ' ', text)   # URLs
    text = re.sub(r'\S+@\S+', ' ', text)            # emails
    text = re.sub(r'<.*?>', ' ', text)               # HTML tags
    text = re.sub(r'[^a-z\s]', ' ', text)            # special chars, digits, emojis
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['clean_text_advanced'] = df['raw_text'].apply(advanced_clean)
print(df[['raw_text', 'clean_text_advanced']].head(5))
```

### Task 4: Handling Stopwords


```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    tokens = word_tokenize(text)
    tokens = [w for w in tokens if w not in stop_words and len(w) > 1]
    return ' '.join(tokens)

df['text_no_stopwords'] = df['clean_text_advanced'].apply(remove_stopwords)
print(df[['clean_text_advanced', 'text_no_stopwords']].head(5))
```

### Task 5: Repeated Characters & Slang (Optional)


```python
slang_dict = {
    'u': 'you',
    'gr8': 'great',
    'soooo': 'so',
    'ur': 'your',
    'thx': 'thanks'
}

def normalize_text(text):
    # repeated characters: soooo -> so
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    tokens = text.split()
    tokens = [slang_dict.get(t, t) for t in tokens]
    return ' '.join(tokens)

df['text_normalized'] = df['text_no_stopwords'].apply(normalize_text)
print('Example slang/repeat fix:')
print('before:', df['text_no_stopwords'].iloc[24])
print('after :', df['text_normalized'].iloc[24])
```

## PART 3 — Basic Text Preprocessing in NLP

### Task 6: Tokenization


```python
print('Word and sentence tokens for 3 samples:\n')
for i in [0, 5, 24]:
    text = df['raw_text'].iloc[i]
    print('--- Sample', i, '---')
    print('Text:', text)
    print('Sentences:', sent_tokenize(text))
    print('Words:', word_tokenize(text.lower()))
    print()
```

### Task 7: Stemming


```python
stemmer = PorterStemmer()

def stem_text(text):
    tokens = word_tokenize(text)
    return ' '.join(stemmer.stem(t) for t in tokens)

df['stemmed_text'] = df['text_no_stopwords'].apply(stem_text)

print('Original vs Stemmed words:')
for w in ['running', 'studies', 'connected', 'phones', 'amazing', 'better']:
    print(w, '->', stemmer.stem(w))

print('\nSample rows:')
print(df[['text_no_stopwords', 'stemmed_text']].head(3))
```

### Task 8: Lemmatization


```python
lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    tokens = word_tokenize(text)
    return ' '.join(lemmatizer.lemmatize(t) for t in tokens)

df['lemmatized_text'] = df['text_no_stopwords'].apply(lemmatize_text)

print('Stemming vs Lemmatization:')
for w in ['running', 'studies', 'better', 'cars', 'was']:
    print(f'{w}: stem={stemmer.stem(w)}, lemma={lemmatizer.lemmatize(w)}, verb_lemma={lemmatizer.lemmatize(w, pos="v")}')

print('\nSample rows:')
print(df[['stemmed_text', 'lemmatized_text']].head(3))
```

### Task 9: Final NLP Pipeline


```python
def nlp_preprocess(text):
    # 1. lowercase
    text = text.lower()
    # 2. noise removal
    text = re.sub(r'http\S+|www\.\S+', ' ', text)
    text = re.sub(r'\S+@\S+', ' ', text)
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    # 3. tokenize + stopwords
    tokens = word_tokenize(text)
    tokens = [w for w in tokens if w not in stop_words and len(w) > 1]
    # 4. normalize repeats + slang
    tokens = [re.sub(r'(.)\1{2,}', r'\1', t) for t in tokens]
    tokens = [slang_dict.get(t, t) for t in tokens]
    # 5. lemmatization
    tokens = [lemmatizer.lemmatize(t) for t in tokens]
    return ' '.join(tokens)

df['final_clean_text'] = df['raw_text'].apply(nlp_preprocess)
print(df[['raw_text', 'final_clean_text']].head(5))
```

### Task 10: Observations & Insights


```python
print('''
1. Basic vs advanced cleaning:
   Basic only lowercases and removes punctuation/numbers.
   Advanced also removes URLs, emails, HTML and special characters.
   Advanced is better for web/review/social text.

2. Why lemmatization is preferred over stemming:
   Stemming cuts word endings and can create non-words (studies -> studi).
   Lemmatization returns real dictionary words (studies -> study).
   Better for meaning and later NLP tasks.

3. Importance of preprocessing in NLP:
   Noise and stopwords add useless tokens.
   Clean text makes models focus on real content.
   A single pipeline keeps train and test text cleaned the same way.
''')

df.to_csv('nlp_preprocessed.csv', index=False)
print('Saved nlp_preprocessed.csv')
print('Columns:', df.columns.tolist())
```
